```python
from langchain.messages import AIMessage
from IPython.display import display, HTML
import math
import html


def render_message_with_prob(message: AIMessage):
    power = 0.6
    max_alpha = 0.60
    metadata = message.response_metadata
    tokens_descriptors = metadata["logprobs"]["content"]
    spans = []
    for token_descriptor in tokens_descriptors:
        token = token_descriptor["token"]
        logprob = token_descriptor["logprob"]
        prob = math.exp(logprob)
        uncertainty = 1 - prob
        uncertainty = uncertainty ** power
        alpha = max_alpha * uncertainty
        token_safe = html.escape(token)
        title = f"Prob: {prob:.4f}"
        if "top_logprobs" in token_descriptor:
            if len(token_descriptor["top_logprobs"]) > 1:
                next_possible_token_descriptor = token_descriptor["top_logprobs"][1]
                next_logprob = next_possible_token_descriptor["logprob"]
                next_prob = math.exp(next_logprob)
                title += f", Next: {next_possible_token_descriptor['token']} (Prob: {next_prob:.4f})"
        style = f"background: rgba(255, 230, 150, {alpha:.3f});"
        spans.append(f"<span style='{style}' title='{title}'>{token_safe}</span>")
    return "<div style='font-family: ui-monospace, SFMono-Regular, Menlo, monospace; line-height: 1.6; white-space: pre-wrap;'>" + "".join(spans) + "</div>"
        
```
